```python
import json
from collections import defaultdict

def analyze_students(file_path):
    with open(file_path) as f:
        data = json.load(f)

    cleaned = []
    seen_ids = set()

    for s in data:
  
        if s.get("id") in seen_ids:
            continue
        seen_ids.add(s.get("id"))

 
        name = (s.get("name") or "Unknown").strip()   

   
        course_raw = (s.get("course") or "").strip().lower()
        course_map = {
            "web dev": "Web Development",
            "web development": "Web Development",
            "data science": "Data Science",
            "ai basics": "AI Basics",
            "cyber security": "Cyber Security"
        } 
        course = course_map.get(course_raw, course_raw.title())

       
        try:
            hours = int(s.get("study_hours"))
        except:
            continue  

      
        try:
            score = int(s.get("score"))
        except:
            score = None

        cleaned.append({
            "id": s["id"],
            "name": name,
            "course": course,
            "study_hours": hours,
            "score": score
        })


    total_hours = sum(s["study_hours"] for s in cleaned) 
    avg_hours = total_hours / len(cleaned)

    valid_scores = [s["score"] for s in cleaned if s["score"] is not None]
    avg_score = sum(valid_scores) / len(valid_scores)

    top_student = max(cleaned, key=lambda x: x["study_hours"])

  
    course_count = defaultdict(int)
    for s in cleaned:
        course_count[s["course"]] += 1


    print("\n✅ CLEANED RECORDS:", len(cleaned))
    print("📊 Average Study Hours:", round(avg_hours, 2))
    print("📊 Average Score:", round(avg_score, 2))
    print("🏆 Most Dedicated Student:", top_student["name"])
    print("\n📚 Students Per Course:")

    for c, count in course_count.items():
        print(f"   {c}: {count}")

    print("\n⚠️ Low Engagement Students:")
    for s in cleaned:
        if s["study_hours"] < avg_hours:
            print(f"   {s['name']} ({s['study_hours']} hrs)")

analyze_students("mydata.json")
```
